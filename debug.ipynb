```python
from pathlib import Path
from hydra import initialize_config_dir, compose
from omegaconf import OmegaConf, open_dict

ROOT = Path("/workspace/code/diffusion-forcing")
CONF_DIR = ROOT / "configurations"

with initialize_config_dir(str(CONF_DIR), job_name="debug"):
    cfg = compose(config_name="config", overrides=[
        "experiment=exp_planning",
        "algorithm=rff_planning",
        "dataset=maze2d_umaze",
        "experiment.training.batch_size=8",
    ])
if "_name" not in cfg.experiment:
    with open_dict(cfg):
        cfg.experiment._name = "exp_planning"
        cfg.algorithm._name = "rff_planning"
        cfg.dataset._name = "maze2d_umaze"
```
